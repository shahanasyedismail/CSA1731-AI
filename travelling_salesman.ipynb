```python
from itertools import permutations


def tour_cost(dist, tour):
    cost = 0
    for i in range(len(tour) - 1):
        cost += dist[tour[i]][tour[i + 1]]
    return cost


def tsp_brute_force(dist, start=0):
    n = len(dist)
    cities = [c for c in range(n) if c != start]
    best_cost, best_tour = float("inf"), None

    for perm in permutations(cities):
        tour = [start] + list(perm) + [start]
        cost = tour_cost(dist, tour)
        if cost < best_cost:
            best_cost, best_tour = cost, tour

    return best_cost, best_tour


def tsp_nearest_neighbour(dist, start=0):
    n = len(dist)
    visited = {start}
    tour = [start]
    cur = start

    while len(visited) < n:
        nxt = min((c for c in range(n) if c not in visited),
                  key=lambda c: dist[cur][c])
        visited.add(nxt)
        tour.append(nxt)
        cur = nxt

    tour.append(start)
    return tour_cost(dist, tour), tour


def main():
    n = int(input("Number of cities: "))
    print(f"Enter {n}x{n} distance matrix (row by row, space separated):")
    dist = []
    for i in range(n):
        row = list(map(int, input(f"Row {i}: ").split()))
        if len(row) != n:
            print("Invalid row length")
            return
        dist.append(row)

    start = int(input(f"Start city (0-{n - 1}): "))
    if not 0 <= start < n:
        print("Invalid start city")
        return

    cost, tour = tsp_brute_force(dist, start)
    print("\nOptimal (brute force):")
    print("  Path :", " -> ".join(map(str, tour)))
    print("  Cost :", cost)

    cost, tour = tsp_nearest_neighbour(dist, start)
    print("\nNearest neighbour (heuristic):")
    print("  Path :", " -> ".join(map(str, tour)))
    print("  Cost :", cost)


if __name__ == "__main__":
    main()
```
