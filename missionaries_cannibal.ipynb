```python
from collections import deque


def is_safe(m, c, total_m, total_c):
    # m, c = missionaries / cannibals on the LEFT bank
    if m < 0 or c < 0 or m > total_m or c > total_c:
        return False
    if m > 0 and c > m:                 # left bank unsafe
        return False
    rm, rc = total_m - m, total_c - c
    if rm > 0 and rc > rm:              # right bank unsafe
        return False
    return True


def solve(total_m, total_c, boat_cap):
    start = (total_m, total_c, 0)       # boat = 0 (left), 1 (right)
    goal = (0, 0, 1)

    # all possible boat loads (missionaries, cannibals)
    loads = [(dm, dc)
             for dm in range(boat_cap + 1)
             for dc in range(boat_cap - dm + 1)
             if 1 <= dm + dc <= boat_cap]

    parent = {start: (None, None)}
    queue = deque([start])

    while queue:
        state = queue.popleft()
        if state == goal:
            break

        m, c, boat = state
        for dm, dc in loads:
            if boat == 0:
                nxt = (m - dm, c - dc, 1)
            else:
                nxt = (m + dm, c + dc, 0)

            if nxt in parent:
                continue
            if not is_safe(nxt[0], nxt[1], total_m, total_c):
                continue

            parent[nxt] = (state, (dm, dc))
            queue.append(nxt)

    if goal not in parent:
        return None

    # rebuild path
    path = []
    cur = goal
    while cur is not None:
        prev, move = parent[cur]
        path.append((cur, move))
        cur = prev
    path.reverse()
    return path


def main():
    total_m = int(input("Number of missionaries: "))
    total_c = int(input("Number of cannibals: "))
    boat_cap = int(input("Boat capacity: "))

    path = solve(total_m, total_c, boat_cap)

    if path is None:
        print("No solution possible")
        return

    print(f"\nSolved in {len(path) - 1} crossings:\n")
    print(f"{'Move':<28}{'Left (M,C)':<14}{'Right (M,C)':<14}Boat")
    print("-" * 64)

    for i, ((m, c, boat), move) in enumerate(path):
        if move is None:
            desc = "Start"
        else:
            direction = "->" if boat == 1 else "<-"
            desc = f"{i}. {move[0]}M {move[1]}C {direction}"
        side = "Right" if boat == 1 else "Left"
        print(f"{desc:<28}({m},{c}){'':<9}({total_m - m},{total_c - c}){'':<9}{side}")


if __name__ == "__main__":
    main()
```
