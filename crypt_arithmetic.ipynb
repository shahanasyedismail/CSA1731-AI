```python
from itertools import permutations


def to_num(word, mapping):
    n = 0
    for ch in word:
        n = n * 10 + mapping[ch]
    return n


def solve(words, result):
    all_words = words + [result]
    letters = sorted(set("".join(all_words)))

    if len(letters) > 10:
        return []

    # first letter of a multi-digit word can't be 0
    leading = {w[0] for w in all_words if len(w) > 1}

    solutions = []
    for perm in permutations(range(10), len(letters)):
        mapping = dict(zip(letters, perm))

        if any(mapping[c] == 0 for c in leading):
            continue

        if sum(to_num(w, mapping) for w in words) == to_num(result, mapping):
            solutions.append(mapping)

    return solutions


def main():
    puzzle = input("Enter puzzle (e.g. SEND + MORE = MONEY): ").upper()

    left, right = puzzle.split("=")
    words = [w.strip() for w in left.split("+")]
    result = right.strip()

    solutions = solve(words, result)

    if not solutions:
        print("No solution")
        return

    print(f"\nFound {len(solutions)} solution(s)\n")
    for mapping in solutions:
        print("Mapping:", ", ".join(f"{k}={v}" for k, v in mapping.items()))
        nums = [to_num(w, mapping) for w in words]
        print("  " + " + ".join(str(n) for n in nums) + " = " + str(to_num(result, mapping)))


if __name__ == "__main__":
    main()
```
