```python
def is_consistent(region, color, assignment, neighbors):
    for nb in neighbors[region]:
        if assignment.get(nb) == color:
            return False
    return True


def select_variable(regions, assignment, neighbors, domains):
    # MRV: pick unassigned region with fewest legal colors left
    unassigned = [r for r in regions if r not in assignment]
    return min(unassigned, key=lambda r: sum(
        1 for c in domains[r] if is_consistent(r, c, assignment, neighbors)))


def backtrack(assignment, regions, neighbors, domains):
    if len(assignment) == len(regions):
        return assignment

    region = select_variable(regions, assignment, neighbors, domains)
    for color in domains[region]:
        if is_consistent(region, color, assignment, neighbors):
            assignment[region] = color
            result = backtrack(assignment, regions, neighbors, domains)
            if result is not None:
                return result
            del assignment[region]  # backtrack
    return None


if __name__ == "__main__":
    # Australia map coloring
    regions = ['WA', 'NT', 'SA', 'Q', 'NSW', 'V', 'T']
    colors = ['Red', 'Green', 'Blue']

    neighbors = {
        'WA':  ['NT', 'SA'],
        'NT':  ['WA', 'SA', 'Q'],
        'SA':  ['WA', 'NT', 'Q', 'NSW', 'V'],
        'Q':   ['NT', 'SA', 'NSW'],
        'NSW': ['Q', 'SA', 'V'],
        'V':   ['SA', 'NSW'],
        'T':   []
    }
    domains = {r: colors[:] for r in regions}

    solution = backtrack({}, regions, neighbors, domains)

    if solution is None:
        print("No solution")
    else:
        for r in regions:
            print(f"{r}: {solution[r]}")
```
