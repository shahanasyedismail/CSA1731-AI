```python
def show(step, action, loc, status):
    print(f"{step:<6}{action:<14}{loc:<10}A={status['A']:<8}B={status['B']}")


def vacuum_cleaner(location, status):
    step = 0
    print(f"\n{'Step':<6}{'Action':<14}{'Location':<10}Rooms")
    print("-" * 44)
    show(step, "Start", location, status)

    while status["A"] == "Dirty" or status["B"] == "Dirty":
        step += 1
        if status[location] == "Dirty":
            status[location] = "Clean"
            action = "Suck"
        elif location == "A":
            location = "B"
            action = "Move Right"
        else:
            location = "A"
            action = "Move Left"
        show(step, action, location, status)

    print(f"\nBoth rooms clean in {step} actions")


def main():
    location = input("Starting room (A/B): ").strip().upper()
    a = input("Room A (Dirty/Clean): ").strip().capitalize()
    b = input("Room B (Dirty/Clean): ").strip().capitalize()

    if location not in ("A", "B") or a not in ("Dirty", "Clean") or b not in ("Dirty", "Clean"):
        print("Invalid input")
        return

    vacuum_cleaner(location, {"A": a, "B": b})


if __name__ == "__main__":
    main()
```
