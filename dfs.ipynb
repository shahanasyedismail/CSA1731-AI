```python
def dfs_recursive(graph, node, visited=None, order=None):
    if visited is None:
        visited, order = set(), []
    visited.add(node)
    order.append(node)
    for nbr in graph[node]:
        if nbr not in visited:
            dfs_recursive(graph, nbr, visited, order)
    return order


def dfs_iterative(graph, start):
    visited, order = set(), []
    stack = [start]
    while stack:
        node = stack.pop()
        if node in visited:
            continue
        visited.add(node)
        order.append(node)
        # reversed so neighbours are visited in listed order
        for nbr in reversed(graph[node]):
            if nbr not in visited:
                stack.append(nbr)
    return order


def dfs_path(graph, start, goal, visited=None):
    if visited is None:
        visited = set()
    visited.add(start)
    if start == goal:
        return [start]
    for nbr in graph[start]:
        if nbr not in visited:
            sub = dfs_path(graph, nbr, goal, visited)
            if sub:
                return [start] + sub
    return None


def main():
    graph = {
        "A": ["B", "C"],
        "B": ["D", "E"],
        "C": ["F"],
        "D": [],
        "E": ["F"],
        "F": [],
    }

    start = input("Start node: ").strip().upper()
    goal = input("Goal node: ").strip().upper()

    if start not in graph or goal not in graph:
        print("Invalid node")
        return

    print("\nRecursive DFS :", " -> ".join(dfs_recursive(graph, start)))
    print("Iterative DFS :", " -> ".join(dfs_iterative(graph, start)))

    path = dfs_path(graph, start, goal)
    if path:
        print(f"Path {start} to {goal}  :", " -> ".join(path))
    else:
        print(f"No path from {start} to {goal}")


if __name__ == "__main__":
    main()
```
