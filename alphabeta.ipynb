```python
import math

leaves_visited = 0


def alphabeta(node, depth, alpha, beta, is_max):
    global leaves_visited

    # Leaf node: return its score
    if depth == 0 or not isinstance(node, list):
        leaves_visited += 1
        return node

    if is_max:
        best = -math.inf
        for child in node:
            best = max(best, alphabeta(child, depth - 1, alpha, beta, False))
            alpha = max(alpha, best)
            if beta <= alpha:
                print(f"  Pruned at MAX node (alpha={alpha}, beta={beta})")
                break
        return best
    else:
        best = math.inf
        for child in node:
            best = min(best, alphabeta(child, depth - 1, alpha, beta, True))
            beta = min(beta, best)
            if beta <= alpha:
                print(f"  Pruned at MIN node (alpha={alpha}, beta={beta})")
                break
        return best


if __name__ == "__main__":
    # Root = MAX, next level = MIN, next = MAX, then leaves
    tree = [
        [[3, 5], [6, 9]],
        [[1, 2], [0, -1]]
    ]

    result = alphabeta(tree, 3, -math.inf, math.inf, True)

    print("Best value:", result)
    print("Leaves visited:", leaves_visited, "out of 8")
```
