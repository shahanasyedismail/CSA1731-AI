```python
from collections import deque

GOAL = (1, 2, 3, 4, 5, 6, 7, 8, 0)  # 0 = blank


def is_solvable(state):
    nums = [x for x in state if x != 0]
    inv = sum(1 for i in range(len(nums))
              for j in range(i + 1, len(nums)) if nums[i] > nums[j])
    return inv % 2 == 0  # even inversions = solvable (3x3)


def get_neighbors(state):
    result = []
    z = state.index(0)
    row, col = divmod(z, 3)
    moves = [('Up', -3, row > 0), ('Down', 3, row < 2),
             ('Left', -1, col > 0), ('Right', 1, col < 2)]
    for name, delta, valid in moves:
        if valid:
            s = list(state)
            s[z], s[z + delta] = s[z + delta], s[z]
            result.append((name, tuple(s)))
    return result


def solve_bfs(start):
    if not is_solvable(start):
        return None
    queue = deque([start])
    parent = {start: (None, None)}  # state -> (prev_state, move)

    while queue:
        state = queue.popleft()
        if state == GOAL:
            path = []
            while parent[state][0] is not None:
                prev, move = parent[state]
                path.append((move, state))
                state = prev
            path.reverse()
            return path
        for move, nxt in get_neighbors(state):
            if nxt not in parent:
                parent[nxt] = (state, move)
                queue.append(nxt)
    return None


def print_board(state):
    for i in range(0, 9, 3):
        print(' '.join(str(x) if x != 0 else '_' for x in state[i:i + 3]))
    print()


if __name__ == "__main__":
    start = (1, 2, 3,
             4, 0, 6,
             7, 5, 8)

    print("Start:")
    print_board(start)

    path = solve_bfs(start)
    if path is None:
        print("No solution (unsolvable puzzle)")
    else:
        print("Moves:", [m for m, _ in path])
        print("Total moves:", len(path), "\n")
        for i, (move, state) in enumerate(path, 1):
            print(f"Step {i}: {move}")
            print_board(state)
```
