```python
from collections import deque
from math import gcd


def water_jug(cap_a, cap_b, target):
    # solvable only if target <= max jug and divisible by gcd
    if target > max(cap_a, cap_b) or target % gcd(cap_a, cap_b) != 0:
        return None

    start = (0, 0)
    visited = {start}
    queue = deque([(start, [start])])

    while queue:
        (a, b), path = queue.popleft()

        if a == target or b == target:
            return path

        # pour amount for transfers
        a_to_b = min(a, cap_b - b)
        b_to_a = min(b, cap_a - a)

        next_states = [
            (cap_a, b),                 # fill A
            (a, cap_b),                 # fill B
            (0, b),                     # empty A
            (a, 0),                     # empty B
            (a - a_to_b, b + a_to_b),   # pour A -> B
            (a + b_to_a, b - b_to_a),   # pour B -> A
        ]

        for state in next_states:
            if state not in visited:
                visited.add(state)
                queue.append((state, path + [state]))

    return None


def describe_step(prev, curr, cap_a, cap_b):
    pa, pb = prev
    ca, cb = curr
    if ca == cap_a and pa != cap_a and cb == pb:
        return "Fill Jug A"
    if cb == cap_b and pb != cap_b and ca == pa:
        return "Fill Jug B"
    if ca == 0 and cb == pb:
        return "Empty Jug A"
    if cb == 0 and ca == pa:
        return "Empty Jug B"
    if ca < pa:
        return "Pour A -> B"
    return "Pour B -> A"


def main():
    cap_a = int(input("Capacity of Jug A: "))
    cap_b = int(input("Capacity of Jug B: "))
    target = int(input("Target amount: "))

    path = water_jug(cap_a, cap_b, target)

    if path is None:
        print("No solution possible")
        return

    print(f"\nSolved in {len(path) - 1} steps:\n")
    print(f"Start        -> (A={path[0][0]}, B={path[0][1]})")
    for i in range(1, len(path)):
        action = describe_step(path[i - 1], path[i], cap_a, cap_b)
        print(f"{action:<13}-> (A={path[i][0]}, B={path[i][1]})")


if __name__ == "__main__":
    main()
```
