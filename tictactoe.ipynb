```python
WIN_LINES = [(0, 1, 2), (3, 4, 5), (6, 7, 8),
             (0, 3, 6), (1, 4, 7), (2, 5, 8),
             (0, 4, 8), (2, 4, 6)]

HUMAN, AI = 'X', 'O'


def winner(board):
    for a, b, c in WIN_LINES:
        if board[a] != ' ' and board[a] == board[b] == board[c]:
            return board[a]
    return None


def minimax(board, depth, is_ai, alpha, beta):
    w = winner(board)
    if w == AI:
        return 10 - depth      # faster win = better
    if w == HUMAN:
        return depth - 10      # slower loss = better
    if ' ' not in board:
        return 0               # draw

    if is_ai:
        best = -100
        for i in range(9):
            if board[i] == ' ':
                board[i] = AI
                best = max(best, minimax(board, depth + 1, False, alpha, beta))
                board[i] = ' '
                alpha = max(alpha, best)
                if beta <= alpha:
                    break      # prune
        return best
    else:
        best = 100
        for i in range(9):
            if board[i] == ' ':
                board[i] = HUMAN
                best = min(best, minimax(board, depth + 1, True, alpha, beta))
                board[i] = ' '
                beta = min(beta, best)
                if beta <= alpha:
                    break      # prune
        return best


def best_move(board):
    best_score, move = -100, None
    for i in range(9):
        if board[i] == ' ':
            board[i] = AI
            score = minimax(board, 1, False, -100, 100)
            board[i] = ' '
            if score > best_score:
                best_score, move = score, i
    return move


def print_board(board):
    for r in range(0, 9, 3):
        print(' | '.join(board[r:r + 3]))
        if r < 6:
            print('--+---+--')
    print()


def get_human_move(board):
    while True:
        try:
            pos = int(input("Your move (1-9): ")) - 1
            if 0 <= pos <= 8 and board[pos] == ' ':
                return pos
            print("Invalid or taken cell, try again.")
        except ValueError:
            print("Enter a number 1-9.")


def play():
    board = [' '] * 9
    print("You = X, AI = O. Cells are numbered:")
    print("1 | 2 | 3\n--+---+--\n4 | 5 | 6\n--+---+--\n7 | 8 | 9\n")

    while True:
        board[get_human_move(board)] = HUMAN
        print_board(board)
        if winner(board) or ' ' not in board:
            break

        print("AI is thinking...")
        board[best_move(board)] = AI
        print_board(board)
        if winner(board) or ' ' not in board:
            break

    w = winner(board)
    if w == HUMAN:
        print("You win!")
    elif w == AI:
        print("AI wins!")
    else:
        print("It's a draw!")


if __name__ == "__main__":
    play()
```
