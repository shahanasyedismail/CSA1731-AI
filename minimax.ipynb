```python
import math


def minimax(state, depth, is_max, is_terminal, evaluate, get_moves, apply_move):
    # Base case: game over or depth limit reached
    if depth == 0 or is_terminal(state):
        return evaluate(state)

    if is_max:
        best = -math.inf
        for move in get_moves(state):
            score = minimax(apply_move(state, move), depth - 1, False,
                            is_terminal, evaluate, get_moves, apply_move)
            best = max(best, score)
        return best
    else:
        best = math.inf
        for move in get_moves(state):
            score = minimax(apply_move(state, move), depth - 1, True,
                            is_terminal, evaluate, get_moves, apply_move)
            best = min(best, score)
        return best


def minimax_ab(state, depth, alpha, beta, is_max,
               is_terminal, evaluate, get_moves, apply_move):
    # Same as minimax but with alpha-beta pruning
    if depth == 0 or is_terminal(state):
        return evaluate(state)

    if is_max:
        best = -math.inf
        for move in get_moves(state):
            score = minimax_ab(apply_move(state, move), depth - 1, alpha, beta, False,
                               is_terminal, evaluate, get_moves, apply_move)
            best = max(best, score)
            alpha = max(alpha, best)
            if beta <= alpha:
                break  # prune
        return best
    else:
        best = math.inf
        for move in get_moves(state):
            score = minimax_ab(apply_move(state, move), depth - 1, alpha, beta, True,
                               is_terminal, evaluate, get_moves, apply_move)
            best = min(best, score)
            beta = min(beta, best)
            if beta <= alpha:
                break  # prune
        return best


if __name__ == "__main__":
    # Game tree as nested lists: leaves are scores
    # Depth 3, MAX at root
    tree = [
        [[3, 5], [6, 9]],
        [[1, 2], [0, -1]]
    ]

    def is_terminal(s):
        return not isinstance(s, list)

    def evaluate(s):
        return s

    def get_moves(s):
        return range(len(s))

    def apply_move(s, i):
        return s[i]

    plain = minimax(tree, 3, True, is_terminal, evaluate, get_moves, apply_move)
    pruned = minimax_ab(tree, 3, -math.inf, math.inf, True,
                        is_terminal, evaluate, get_moves, apply_move)

    print("Minimax value:", plain)
    print("Minimax + alpha-beta value:", pruned)
```
